```
# ================================
# COMPLETE CASSANDRA SETUP - COLAB
# ================================

# 1) Install Java 11 + required packages
!apt-get update -y
!apt-get install -y openjdk-11-jdk curl gnupg

# 2) Add Cassandra repository
!mkdir -p /etc/apt/keyrings

!curl -o /etc/apt/keyrings/apache-cassandra.asc \
https://downloads.apache.org/cassandra/KEYS

!echo "deb [signed-by=/etc/apt/keyrings/apache-cassandra.asc] https://debian.cassandra.apache.org 41x main" \
| tee /etc/apt/sources.list.d/cassandra.sources.list

# 3) Install Cassandra
!apt-get update -y
!apt-get install -y cassandra

# 4) Install newer cqlsh for Colab Python
!pip install -U cqlsh

# 5) Set Java 11 for Cassandra
import os

os.environ["JAVA_HOME"] = "/usr/lib/jvm/java-11-openjdk-amd64"
os.environ["PATH"] = os.environ["JAVA_HOME"] + "/bin:" + os.environ["PATH"]

# Check Java
!$JAVA_HOME/bin/java -version

# 6) Stop any partially started Cassandra process
!pkill -f CassandraDaemon || true

# 7) Start Cassandra directly
!JAVA_HOME=/usr/lib/jvm/java-11-openjdk-amd64 \
PATH=/usr/lib/jvm/java-11-openjdk-amd64/bin:$PATH \
cassandra -R

# Give Cassandra time to start
import time
time.sleep(15)

# 8) Check Cassandra status
!JAVA_HOME=/usr/lib/jvm/java-11-openjdk-amd64 \
PATH=/usr/lib/jvm/java-11-openjdk-amd64/bin:$PATH \
nodetool status

# 9) Check port 9042
!ss -ltn | grep 9042 || true

# 10) Verify cqlsh connection
!/usr/local/bin/cqlsh 127.0.0.1 9042 -e "DESCRIBE KEYSPACES;"

# 11) Create keyspace
!/usr/local/bin/cqlsh 127.0.0.1 9042 -e "CREATE KEYSPACE IF NOT EXISTS training WITH replication = {'class':'SimpleStrategy','replication_factor':1};"

# 12) Create table
!/usr/local/bin/cqlsh 127.0.0.1 9042 -e "CREATE TABLE IF NOT EXISTS training.employee (id int PRIMARY KEY, name text, city text, salary int);"

# 13) Insert sample records
!/usr/local/bin/cqlsh 127.0.0.1 9042 -e "INSERT INTO training.employee (id,name,city,salary) VALUES (1,'Sai','Hyderabad',100000);"

!/usr/local/bin/cqlsh 127.0.0.1 9042 -e "INSERT INTO training.employee (id,name,city,salary) VALUES (2,'Vijay','Bangalore',80000);"

!/usr/local/bin/cqlsh 127.0.0.1 9042 -e "INSERT INTO training.employee (id,name,city,salary) VALUES (3,'Ravi','Chennai',70000);"

# 14) Query Cassandra
!/usr/local/bin/cqlsh 127.0.0.1 9042 -e "SELECT * FROM training.employee;"

!pip install -q pyspark==3.5.1
```
