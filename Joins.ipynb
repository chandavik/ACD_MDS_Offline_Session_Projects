```python
data4 = [
    (1, "raj"),
    (2, "ravi"),
    (3, "sai"),
    (5, "rani")
]

cust = spark.createDataFrame(data4, ["id", "name"])
cust.show()

data3 = [
    (1, "mouse"),
    (3, "mobile"),
    (7, "laptop")
]

prod = spark.createDataFrame(data3, ["id", "product"])
prod.show()

id_list = [row.id for row in prod.select("id").collect()]
print(id_list)


df_filter = cust.filter( ~col("id").isin(id_list))
df_filter.show()



print("===== ANTI_JOIN======")
print()
antijoin = cust.join(prod , ["id"] , "left_anti")
antijoin.show()




innerj = cust.join(prod, ["id"], "inner")
innerj.show()


fullj  = cust.join(prod, ["id"] , "full")
fullj.show()


print("===== LEFT_ANTI ======")
print()

fullj.join( innerj , ["id"] , "left_anti").show()

print("===== SUBTRACT ======")
print()

fullj.subtract(innerj).show()
```

```python
# AGG



data = [
    ("sai",  "chn", 10),
    ("zeyo", "hyd", 5),
    ("sai",  "hyd", 5),
    ("zeyo", "chn", 5),
    ("sai",  "hyd", 10),
    ("zeyo",  "hyd", 10),
    ("sai",  "chn", 10)
]

df = spark.createDataFrame(data).toDF("name", "city", "amt")

df.show()

from pyspark.sql.functions import *


aggdf = (

        df.groupBy("name")
          .agg(

            sum("amt").alias("total"),
            count("amt").alias("cnt"),
            collect_list("amt").alias("amt_collect")

        )



)

aggdf.show()
```

```python
data = [
    ("sai",  "chn", 10),
    ("zeyo", "hyd", 5),
    ("sai",  "hyd", 5),
    ("zeyo", "chn", 5),
    ("sai",  "hyd", 10),
    ("zeyo",  "hyd", 10),
    ("sai",  "chn", 10)
]

df = spark.createDataFrame(data).toDF("name", "city", "amt")

df.show()


from pyspark.sql.functions import *


aggdf = df.groupBy("name").agg(

                            sum("amt").alias("total"),

                            count("amt").alias("cnt"),

                            collect_list("amt").alias("amt_collect"),

                            collect_set("amt").alias("amt_distinct")


                    )

aggdf.show()

df.createOrReplaceTempView("df")


spark.sql("""

                select name,
                        sum(amt) as total,
                        count(amt) as cnt,
                        collect_list(amt) as amt_collect,
                        collect_set(amt)  as amt_distinct
                from df
                group by name

                        """).show()
```

```python
data1 = [
    (1, "abc", 31, "abc@gmail.com"),
    (2, "def", 23, "defyahoo.com"),
    (3, "xyz", 26, "xyz@gmail.com"),
    (4, "qwe", 34, "qwegmail.com"),
    (5, "iop", 24, "iop@gmail.com")
]
myschema1 = ["id", "name", "age", "email"]
df1 = spark.createDataFrame(data1, schema=myschema1)
df1.show()

data2 = [
    (11, "jkl", 22, "abc@gmail.com", 1000),
    (12, "vbn", 33, "vbn@yahoo.com", 3000),
    (13, "wer", 27, "wer", 2000),
    (14, "zxc", 30, "zxc.com", 2000),
    (15, "lkj", 29, "lkj@outlook.com", 2000)
]
myschema2 = ["id", "name", "age", "email", "salary"]
df2 = spark.createDataFrame(data2, schema=myschema2)
df2.show()



# 🔴🔴ATTACH 1000 for first table


saldf = df1.withColumn("salary" , expr("1000"))
saldf.show()





# 🔴🔴Union df


uniondf = saldf.union(df2)
uniondf.show()






# 🔴🔴 filter like  '%@%'


fildf = uniondf.filter( "email like '%@%'  ")
fildf.show()
```

```python
# Multi Column Group By

data = [
    ("sai",  "chn", 10),
    ("zeyo", "hyd", 5),
    ("sai",  "hyd", 5),
    ("zeyo", "chn", 5),
    ("sai",  "hyd", 10),
    ("zeyo",  "hyd", 10),
    ("sai",  "chn", 10)
]

df = spark.createDataFrame(data).toDF("name", "city", "amt")

df.show()


from pyspark.sql.functions import *


aggdf = df.groupBy("name","city").agg(

                            sum("amt").alias("total"),

                            count("amt").alias("cnt"),

                            collect_list("amt").alias("amt_collect"),

                            collect_set("amt").alias("amt_distinct")


                    )

aggdf.show()

df.createOrReplaceTempView("df")


spark.sql("""

                select name,city,
                        sum(amt) as total,
                        count(amt) as cnt,
                        collect_list(amt) as amt_collect,
                        collect_set(amt)  as amt_distinct
                from df
                group by name,city

                        """).show()
```
